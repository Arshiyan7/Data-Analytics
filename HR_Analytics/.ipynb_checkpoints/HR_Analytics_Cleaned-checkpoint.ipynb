# Data Cleaning Plan

## Objective
Before moving into exploratory analysis and business questions, the raw HR dataset needs to be cleaned. Initial exploration using `.info()`, `.describe()`, `.isnull().sum()`, and `.nunique()` revealed inconsistencies typical of real-world HR data: whitespace issues, inconsistent text casing, and redundant columns that need to be cross-validated against each other.

## Issues Identified

### 1. Whitespace inconsistencies
- `Sex` — values like `'M '` contain trailing whitespace
- `Department` — values like `'Production       '` contain excessive trailing whitespace
- `Position` — `'Data Analyst'` and `'Data Analyst '` exist as separate values due to trailing whitespace

### 2. Inconsistent text casing
- `HispanicLatino` — contains `'Yes'`, `'No'`, `'yes'`, `'no'` as four separate values instead of two
- `TermReason` — mixed casing across entries (e.g. `'career change'` vs `'Another position'`)

### 3. Data quality anomalies
- `TermReason` contains two entries that appear to be placeholder or joke text (`'Learned that he is a gangster'`, `'Fatal attraction'`) rather than genuine termination reasons. These need to be recoded (e.g. to `'Other'`) rather than deleted, to avoid losing legitimate rows over a labeling issue.

### 4. Column redundancy requiring validation
Three columns describe the same underlying fact — whether an employee is still active or has left the company:
- `Termd` (0/1 flag)
- `EmploymentStatus` (Active / Voluntarily Terminated / Terminated for Cause)
- `TermReason` (text reason, or `'N/A-StillEmployed'` if active)

Since these columns should logically agree with each other, a crosstab check is used to confirm there are no contradictions (e.g. an employee marked `Termd = 0` but `EmploymentStatus = 'Voluntarily Terminated'`).

## Planned Actions
1. Strip whitespace from `Sex`, `Department`, and `Position`
2. Standardize casing on `HispanicLatino` to two consistent values
3. Standardize casing on `TermReason` and recode placeholder entries to `'Other'`
4. Run `pd.crosstab(df['EmploymentStatus'], df['Termd'])` to confirm status and termination flag agree
5. Run a check confirming `TermReason` correctly shows `'N/A-StillEmployed'` for every `Active` employee

```python
df['Sex'] = df['Sex'].str.strip()
```

```python
df['HispanicLatino'] = df['HispanicLatino'].str.strip().str.capitalize()
```

```python
df['Department'] = df['Department'].str.strip()
```

```python
df['Position'] = df['Position'].str.strip()
df['Position'].nunique()
```

```python
df['TermReason'] = df['TermReason'].str.strip().str.lower()
df['TermReason'] = df['TermReason'].replace({
    'learned that he is a gangster': 'other',
    'fatal attraction': 'other'
})
```

```python
pd.crosstab(df['EmploymentStatus'], df['Termd'])
```

```python
pd.crosstab(df['EmploymentStatus'], df['TermReason'] == 'N/A-StillEmployed')
```

```python
df['TermReason'].value_counts()
```

# Cleaning Conclusion: Status Consistency Check

## What Was Checked
A crosstab was run between `EmploymentStatus` and `Termd` to confirm the two columns agree on employment status across all 311 records.

## Result

| EmploymentStatus         | Termd = 0 | Termd = 1 |
|---------------------------|-----------|-----------|
| Active                     | 207       | 0         |
| Terminated for Cause       | 0         | 16        |
| Voluntarily Terminated     | 0         | 88        |

The result is a clean diagonal with no contradictions. Every employee marked `Active` has `Termd = 0`, and every terminated employee (for cause or voluntary) has `Termd = 1`. This confirms `Termd` and `EmploymentStatus` are fully consistent with each other, and no rows need correction on this basis.

## Note on TermReason Check
A second check compared `EmploymentStatus` against whether `TermReason` equals `'N/A-StillEmployed'`. The result requires re-verification, as the row counts (207 / 16 / 88) matched the `EmploymentStatus` breakdown exactly but all returned `False`, which is inconsistent with the expectation that all 207 `Active` employees should show `'N/A-StillEmployed'` as `True`. This should be re-run using `df['TermReason'].value_counts()` to directly confirm the count of `'N/A-StillEmployed'` entries before finalizing this check.

## Outcome
`Termd` and `EmploymentStatus` are confirmed consistent. No data correction needed for this pair. The `TermReason` validation is pending re-verification before this column can be considered fully checked.
